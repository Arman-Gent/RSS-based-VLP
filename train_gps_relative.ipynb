```python
import numpy as np
import gpflow
import matplotlib.pyplot as plt
import os
import pandas as pd
import pickle
from sklearn.preprocessing import StandardScaler

# ----------------------------------------------------------------------
# 1) Define your configuration
# ----------------------------------------------------------------------

# Suppose our new folder is "Measured_relative_Normalized"
categories = ['Measured_relative_Normalized']

# For demonstration, let's assume the same single noise_level as before:
noise_levels = [0.0]

# Range of sample sizes, with increments of 50
ns = range(50, 1501, 50)

# We have multiple seeds, e.g. 0..9
seeds = range(10)

base_dir = '3D-Data'  # parent directory containing subfolders

# ----------------------------------------------------------------------
# 2) Load data into a dictionary keyed by (category, noise_level, n, seed)
#    This time, the naming convention is "relative_rss_n={n}_noise={noise_level}_seed={seed}.csv"
# ----------------------------------------------------------------------

train_dict = {}
for category in categories:
    for noise_level in noise_levels:
        for n in ns:
            for seed in seeds:
                file_name = f"relative_rss_n={n}_noise={noise_level}_seed={seed}.csv"
                file_path = os.path.join(base_dir, category, file_name)
                
                if os.path.exists(file_path):
                    df = pd.read_csv(file_path)
                    train_dict[(category, noise_level, n, seed)] = df
                else:
                    print(f"No such file: {file_path}")

# ----------------------------------------------------------------------
# 3) Define a training function that uses the desired feature columns
#    e.g. "RSS0/RSS1", "RSS0/RSS2", etc.
# ----------------------------------------------------------------------
```

```python

def gp_trainer_relative(train_dict):
    """
    Train GPR models on 'relative' data:
    - Each CSV has columns like "RSS0/RSS1", "RSS0/RSS2", etc.
    - We store each trained model keyed by (category, noise_level, n, seed).
    """
    gp_models = {}
    
    for (category, noise_level, n, seed), df in train_dict.items():
        # Example: The columns below might be "RSS0/RSS1", "RSS0/RSS2", ...
        # Adjust if your CSV columns differ.
        feature_cols = [
            "RSS0/RSS1", "RSS0/RSS2", "RSS0/RSS3", "RSS0/RSS4",
            "RSS1/RSS2", "RSS1/RSS3", "RSS1/RSS4",
            "RSS2/RSS3", "RSS2/RSS4",
            "RSS3/RSS4"
        ]
        
        # Extract the input/features and output/labels
        train_input = df[feature_cols].values
        train_output = df[["X", "Y", "Z"]].values

        # Example kernel & noise variance
        kernel = gpflow.kernels.SquaredExponential(
            lengthscales=np.ones(len(feature_cols)) * 0.2
        )
        
        model = gpflow.models.GPR(
            data=(train_input, train_output),
            kernel=kernel,
            noise_variance=1e-4
        )
        
        # Make likelihood trainable, do optimization
        gpflow.set_trainable(model.likelihood, True)
        opt = gpflow.optimizers.Scipy()
        opt.minimize(model.training_loss, model.trainable_variables)
        
        # Store in the dictionary
        gp_models[(category, noise_level, n, seed)] = model
        
        print(f"Trained model for: category={category}, noise={noise_level}, n={n}, seed={seed}")
    
    return gp_models

# ----------------------------------------------------------------------
# 4) Train and save all models
# ----------------------------------------------------------------------

gp_models_relative = gp_trainer_relative(train_dict)

# Save them to disk with a new filename, e.g. "gp_models_relative.pkl"
with open('gp_models_relative.pkl', 'wb') as file:
    pickle.dump(gp_models_relative, file)

print("Finished training all GP 'relative' models (one per seed).")
```

```python
test_dict={}
m=50
noise_levels= [0.,0.00019,1.9e-05,6e-05, 6e-06]
for category in categories:
    for noise_level in noise_levels:
        file_name= f"relative_gnd_n={m}_noise={noise_level}.csv"
        file_path = os.path.join(base_dir,category,file_name)
        if os.path.exists(file_path):
            test_dict[(category,noise_level)] = pd.read_csv(file_path)
        else:
            print('no such file', category)
def make_predictions(
    gp_model,
    test_data,
    prediction_save_directory="predictions/gps_relative",
    n=9,
    noise_level=0.,
    seed=0,
    scaler="scaler_relative.pkl"
):
    """
    Make predictions for the given GP model and test data, then save to CSV.
    Includes `seed` in the output filename to avoid overwriting different models.
    """
    import pickle
    import numpy as np
    import os

    output_columns = ["X", "Y", "Z"]

    # Build a filename that includes seed
    filename = f"rss_n={n}_noise={noise_level}_seed={seed}.csv"

    RSS_test = test_data.drop(output_columns, axis=1).values
    mean, covariance = gp_model.predict_f(RSS_test)

    # Load the relative-data scaler
    with open(scaler, 'rb') as scaler_file:
        loaded_scaler = pickle.load(scaler_file)

    # Descale the predictions
    predictions = mean
    descaled = loaded_scaler.inverse_transform(
        np.hstack([np.zeros_like(RSS_test), predictions])
    )[:, -3:]  # keep only X, Y, Z

    predictions_df = pd.DataFrame(descaled, columns=output_columns)

    # Ensure directory exists
    if not os.path.exists(prediction_save_directory):
        os.makedirs(prediction_save_directory)

    # Save predictions
    prediction_save_path = os.path.join(prediction_save_directory, filename)
    predictions_df.to_csv(prediction_save_path, index=False)
    print(f"Saved predictions to {prediction_save_path}")

    return predictions_df
```

```python
# i is (category, noise_level, n, seed)
for i in gp_models.keys():
    for j in test_dict.keys():
        make_predictions(
            gp_models[i],
            test_dict[j],
            n=i[2],
            noise_level=j[1],
            seed=i[3],  # Pass the seed from the key
            prediction_save_directory="predictions/gps_relative"
        )
```
